## Coding Principles

This notes build heavily on coding, so here are some of my programming principles: 

**Variable names:**     

1. When there are more than one elements in variable name, I use _ to join them. 
    
    *    "a_max" means "maximum of a". I will not use "amax", since I sometimes forget its meaning and may take it as "I am ax".

2. Try to avoid use one letter as abbreviation. Use it only for important variables. 

      *   In consumption saving problem, a is for asset, c is for consumption. Most other variable won't have a one-letter name.

3. Try to use full word, in order to eliminate confusion. As long as I can't get clear meaning of an abbreviation outside the script, I use full word. If I can get the correct meaning by google, I accept the abbreviation.
           
    * When introduce tolerance, "tolerance" is used instead of "tole", since I can't link "tole" to tolerance after several days. And when I google "tole", there is also link to "tolerance".
   
    * When I google "diff", I can know usually it means "difference", so "diff" is accepted. 

        
4.  Use the uppercase for vector and matrix.

     *    In consumption saving problem, A is for asset vector,  V is for value function vector.

5. When there are several variables under one class, put their class name in advance.

      *  for a series of iteration variable, I use "iteration_max" to denote max times of iteration, instead of "max_iteration". 

**Space:**

1. Use a space between operation and numbers.

      * "1 + 1 = 2" is used, instead of "1+1=2".
      * "a = [1, 2, 3]" is used, instead of "a = [1,2,3]".
